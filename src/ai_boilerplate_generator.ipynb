## Core Data Structures

````python
@dataclass
class GenerationRequest:
    """Request for code generation"""
    pattern: str  # 'rest_api', 'crud', 'component', 'class', 'migration'
    framework: str
    language: str
    spec: Dict[str, Any]  # Pattern-specific parameters
    style_guide: Optional[str] = None

@dataclass
class GeneratedCode:
    """Generated code result"""
    code: str
    pattern: str
    framework: str
    language: str
    files: Dict[str, str]  # filename -> code
    explanation: str
    setup_instructions: List[str]
    dependencies: List[str]
    metadata: Dict[str, Any]

@dataclass
class RefactoringResult:
    """Result from refactoring operation"""
    original_code: str
    refactored_code: str
    changes_summary: str
    improvements: List[str]
    breaking_changes: List[str]
    test_plan: str
    metadata: Dict[str, Any]

@dataclass
class TelemetryLog:
    """Usage telemetry"""
    timestamp: datetime
    operation: str  # 'generate', 'refactor', 'convert'
    pattern: str
    framework: str
    model: str
    provider: str
    prompt_tokens: int
    completion_tokens: int
    total_tokens: int
    cost_usd: float
    latency_ms: float
    lines_generated: int

print("✅ Data structures defined")
````

## AI Provider Integration

````python
class AIProvider:
    """Base AI provider"""
    def complete(self, system_prompt: str, user_prompt: str, 
                 temperature: float = 0.3) -> Tuple[str, Dict[str, Any]]:
        raise NotImplementedError

class OpenAIProvider(AIProvider):
    def __init__(self, api_key: str, model: str = 'gpt-5'):
        from openai import OpenAI
        self.client = OpenAI(api_key=api_key)
        self.model = model
    
    def complete(self, system_prompt: str, user_prompt: str, 
                 temperature: float = 0.3) -> Tuple[str, Dict[str, Any]]:
        start = time.time()
        
        response = self.client.chat.completions.create(
            model=self.model,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=temperature,
            max_completion_tokens=3000  # Larger for code generation
        )
        
        latency = (time.time() - start) * 1000
        usage = response.usage
        
        pricing = MODEL_PRICING.get(self.model, MODEL_PRICING['gpt-5'])
        cost = (
            (usage.prompt_tokens / 1_000_000) * pricing['input'] +
            (usage.completion_tokens / 1_000_000) * pricing['output']
        )
        
        metadata = {
            'model': self.model,
            'provider': 'openai',
            'prompt_tokens': usage.prompt_tokens,
            'completion_tokens': usage.completion_tokens,
            'total_tokens': usage.total_tokens,
            'cost_usd': cost,
            'latency_ms': latency
        }
        
        return response.choices[0].message.content, metadata

class AnthropicProvider(AIProvider):
    def __init__(self, api_key: str, model: str = 'claude-3-5-haiku-20241022'):
        from anthropic import Anthropic
        self.client = Anthropic(api_key=api_key)
        self.model = model
    
    def complete(self, system_prompt: str, user_prompt: str, 
                 temperature: float = 0.3) -> Tuple[str, Dict[str, Any]]:
        start = time.time()
        
        response = self.client.messages.create(
            model=self.model,
            system=system_prompt,
            messages=[{"role": "user", "content": user_prompt}],
            temperature=temperature,
            max_completion_tokens=3000
        )
        
        latency = (time.time() - start) * 1000
        
        pricing = MODEL_PRICING.get(self.model, MODEL_PRICING['claude-3-5-haiku-20241022'])
        cost = (
            (response.usage.input_tokens / 1_000_000) * pricing['input'] +
            (response.usage.output_tokens / 1_000_000) * pricing['output']
        )
        
        metadata = {
            'model': self.model,
            'provider': 'anthropic',
            'prompt_tokens': response.usage.input_tokens,
            'completion_tokens': response.usage.output_tokens,
            'total_tokens': response.usage.input_tokens + response.usage.output_tokens,
            'cost_usd': cost,
            'latency_ms': latency
        }
        
        return response.content[0].text, metadata

# Initialize providers
try:
    if OPENAI_API_KEY:
        primary_provider = OpenAIProvider(OPENAI_API_KEY, PRIMARY_MODEL)
        print(f"✅ Primary provider initialized: {PRIMARY_MODEL}")
    else:
        primary_provider = None
        print("⚠️  No OpenAI API key")
    
    if ANTHROPIC_API_KEY:
        fallback_provider = AnthropicProvider(ANTHROPIC_API_KEY, FALLBACK_MODEL)
        print(f"✅ Fallback provider initialized: {FALLBACK_MODEL}")
    else:
        fallback_provider = None
        print("⚠️  No Anthropic API key")
except Exception as e:
    print(f"❌ Error: {e}")
    primary_provider = None
    fallback_provider = None
````

## Boilerplate Generator
Main class for code generation and refactoring

````python
class AIBoilerplateGenerator:
    """AI-powered code generation and refactoring"""
    
    def __init__(self, primary: AIProvider, fallback: Optional[AIProvider] = None):
        self.primary = primary
        self.fallback = fallback
        self.telemetry: List[TelemetryLog] = []
    
    def _call_ai(self, system_prompt: str, user_prompt: str) -> Tuple[str, Dict[str, Any]]:
        """Call AI with fallback"""
        try:
            return self.primary.complete(system_prompt, user_prompt)
        except Exception as e:
            if self.fallback:
                print(f"⚠️  Primary failed, using fallback: {e}")
                return self.fallback.complete(system_prompt, user_prompt)
            raise e
    
    def _log_telemetry(self, operation: str, pattern: str, framework: str,
                      metadata: Dict[str, Any], lines_generated: int):
        """Log usage telemetry"""
        log = TelemetryLog(
            timestamp=datetime.now(),
            operation=operation,
            pattern=pattern,
            framework=framework,
            model=metadata['model'],
            provider=metadata['provider'],
            prompt_tokens=metadata['prompt_tokens'],
            completion_tokens=metadata['completion_tokens'],
            total_tokens=metadata['total_tokens'],
            cost_usd=metadata['cost_usd'],
            latency_ms=metadata['latency_ms'],
            lines_generated=lines_generated
        )
        self.telemetry.append(log)
    
    def generate_rest_api(self, spec: Dict[str, Any], framework: str = 'fastapi') -> GeneratedCode:
        """Generate REST API boilerplate"""
        entity = spec.get('entity', 'Item')
        fields = spec.get('fields', [])
        auth = spec.get('auth', False)
        
        system_prompt = f"""You are an expert {framework} developer.
Generate production-ready REST API code with:
- CRUD operations (GET, POST, PUT, DELETE)
- Input validation
- Error handling
- Type hints
- Docstrings
- Security best practices

Return ONLY the code, properly formatted."""
        
        user_prompt = f"""Generate a {framework} REST API for entity: {entity}

Fields:
{json.dumps(fields, indent=2)}

{'Include authentication and authorization.' if auth else ''}

Generate:
1. Data models/schemas
2. CRUD endpoints
3. Database operations (assume PostgreSQL with SQLAlchemy)
4. Input validation
5. Error handling

Organize into logical files (models.py, routes.py, schemas.py, etc.)"""
        
        response, metadata = self._call_ai(system_prompt, user_prompt)
        
        # Parse response into files
        files = self._parse_code_blocks(response)
        lines = sum(len(code.split('\n')) for code in files.values())
        
        self._log_telemetry('generate', 'rest_api', framework, metadata, lines)
        
        return GeneratedCode(
            code=response,
            pattern='rest_api',
            framework=framework,
            language='python',
            files=files,
            explanation=f"Complete REST API for {entity} with {len(fields)} fields",
            setup_instructions=[
                f"pip install {framework} sqlalchemy pydantic",
                "Create PostgreSQL database",
                "Run migrations",
                f"{framework} dev" if framework == 'fastapi' else "python manage.py runserver"
            ],
            dependencies=['fastapi', 'sqlalchemy', 'pydantic', 'uvicorn'] if framework == 'fastapi' else [],
            metadata=metadata
        )
    
    def generate_react_component(self, spec: Dict[str, Any]) -> GeneratedCode:
        """Generate React component with hooks"""
        component_name = spec.get('name', 'MyComponent')
        props = spec.get('props', [])
        state = spec.get('state', [])
        api_endpoint = spec.get('api_endpoint')
        
        system_prompt = """You are an expert React developer.
Generate modern React components using:
- Functional components with hooks
- TypeScript for type safety
- Proper state management
- Error handling
- Loading states
- Accessibility (ARIA attributes)

Follow React best practices and latest patterns."""
        
        user_prompt = f"""Generate a React component: {component_name}

Props: {json.dumps(props, indent=2)}
State: {json.dumps(state, indent=2)}
{f'Fetch data from: {api_endpoint}' if api_endpoint else ''}

Include:
1. TypeScript interface for props
2. useState and useEffect hooks
3. Error boundary
4. Loading spinner
5. Styled with Tailwind CSS classes
6. Unit test file using Jest and React Testing Library

Return component file and test file separately."""
        
        response, metadata = self._call_ai(system_prompt, user_prompt)
        
        files = self._parse_code_blocks(response)
        lines = sum(len(code.split('\n')) for code in files.values())
        
        self._log_telemetry('generate', 'react_component', 'react', metadata, lines)
        
        return GeneratedCode(
            code=response,
            pattern='react_component',
            framework='react',
            language='typescript',
            files=files,
            explanation=f"React component: {component_name} with hooks and TypeScript",
            setup_instructions=[
                "npm install react react-dom",
                "npm install -D @types/react @types/react-dom",
                "npm install -D jest @testing-library/react @testing-library/jest-dom"
            ],
            dependencies=['react', 'react-dom'],
            metadata=metadata
        )
    
    def generate_apex_class(self, spec: Dict[str, Any]) -> GeneratedCode:
        """Generate Salesforce Apex class"""
        class_name = spec.get('name', 'MyClass')
        sobject = spec.get('sobject', 'Account')
        operations = spec.get('operations', ['create', 'read', 'update', 'delete'])
        trigger = spec.get('trigger', False)
        
        system_prompt = """You are an expert Salesforce Apex developer.
Generate production-ready Apex code with:
- Proper error handling
- Bulkification
- SOQL best practices
- Governor limits awareness
- Security (with sharing)
- Test coverage (>75%)

Follow Salesforce development best practices."""
        
        user_prompt = f"""Generate Apex class: {class_name}

SObject: {sobject}
Operations: {', '.join(operations)}
{'Include trigger handler' if trigger else ''}

Generate:
1. Service class with CRUD methods
2. Bulk-safe operations
3. Error handling
4. Test class with >75% coverage
{'5. Trigger and trigger handler' if trigger else ''}

Return all files separately (class, test, trigger if applicable)."""
        
        response, metadata = self._call_ai(system_prompt, user_prompt)
        
        files = self._parse_code_blocks(response)
        lines = sum(len(code.split('\n')) for code in files.values())
        
        self._log_telemetry('generate', 'apex_class', 'apex', metadata, lines)
        
        return GeneratedCode(
            code=response,
            pattern='apex_class',
            framework='apex',
            language='apex',
            files=files,
            explanation=f"Salesforce Apex class for {sobject} with bulk-safe operations",
            setup_instructions=[
                "Deploy to Salesforce org using SFDX",
                "Run Apex tests: sfdx force:apex:test:run",
                "Verify test coverage is >75%"
            ],
            dependencies=[],
            metadata=metadata
        )
    
    def refactor_legacy_code(self, code: str, language: str, 
                           target_version: str = None) -> RefactoringResult:
        """Refactor legacy code to modern standards"""
        system_prompt = f"""You are an expert {language} refactoring specialist.
Modernize legacy code while preserving functionality:
- Use modern syntax and idioms
- Improve readability and maintainability
- Add type hints (if applicable)
- Fix code smells
- Improve error handling
- Add documentation

List all breaking changes clearly."""
        
        user_prompt = f"""Refactor this legacy {language} code:

```{language}
{code}
```

{f'Target version: {target_version}' if target_version else ''}

Provide:
1. Refactored code
2. Summary of changes
3. List of improvements
4. Breaking changes (if any)
5. Test plan to verify behavior unchanged"""
        
        response, metadata = self._call_ai(system_prompt, user_prompt)
        
        # Extract sections from response
        refactored = self._extract_section(response, 'refactored code', code)
        summary = self._extract_section(response, 'summary', 'Code refactored')
        improvements = self._extract_list(response, 'improvement')
        breaking = self._extract_list(response, 'breaking')
        test_plan = self._extract_section(response, 'test plan', 'Run existing tests')
        
        lines = len(refactored.split('\n'))
        self._log_telemetry('refactor', 'modernization', language, metadata, lines)
        
        return RefactoringResult(
            original_code=code,
            refactored_code=refactored,
            changes_summary=summary,
            improvements=improvements,
            breaking_changes=breaking,
            test_plan=test_plan,
            metadata=metadata
        )
    
    def _parse_code_blocks(self, response: str) -> Dict[str, str]:
        """Extract code blocks from markdown response"""
        files = {}
        pattern = r'```(?:\w+)?\s*(?:#\s*(.+?)\n)?(.*?)```'
        matches = re.findall(pattern, response, re.DOTALL)
        
        for i, (filename, code) in enumerate(matches):
            filename = filename.strip() if filename else f"file_{i+1}.py"
            files[filename] = code.strip()
        
        if not files:
            # No code blocks found, treat entire response as code
            files['main.py'] = response
        
        return files
    
    def _extract_section(self, text: str, section_name: str, default: str) -> str:
        """Extract a section from structured response"""
        pattern = rf'{section_name}[:\s]+(.+?)(?=\n\n|\n[A-Z]|$)'
        match = re.search(pattern, text, re.IGNORECASE | re.DOTALL)
        return match.group(1).strip() if match else default
    
    def _extract_list(self, text: str, keyword: str) -> List[str]:
        """Extract bullet points containing keyword"""
        items = []
        for line in text.split('\n'):
            if keyword.lower() in line.lower() and (line.strip().startswith('-') or line.strip().startswith('•')):
                items.append(line.strip().lstrip('-•').strip())
        return items
    
    def get_telemetry_stats(self) -> Dict[str, Any]:
        """Get usage statistics"""
        if not self.telemetry:
            return {}
        
        df = pd.DataFrame([vars(t) for t in self.telemetry])
        
        return {
            'total_operations': len(df),
            'total_cost': df['cost_usd'].sum(),
            'total_tokens': df['total_tokens'].sum(),
            'total_lines': df['lines_generated'].sum(),
            'avg_latency_ms': df['latency_ms'].mean(),
            'by_operation': df.groupby('operation')['cost_usd'].sum().to_dict(),
            'by_pattern': df.groupby('pattern')['lines_generated'].sum().to_dict(),
            'by_framework': df.groupby('framework')['cost_usd'].sum().to_dict()
        }

# Initialize generator
if primary_provider:
    generator = AIBoilerplateGenerator(primary_provider, fallback_provider)
    print("✅ Boilerplate Generator initialized")
else:
    generator = None
    print("⚠️  Generator not available - configure API keys")
````

## 🚀 Run All Demos in Parallel
Execute all 4 code generation demos simultaneously to save time

````python
if generator:
    from concurrent.futures import ThreadPoolExecutor, as_completed
    import time
    
    print("="*80)
    print("🚀 RUNNING ALL DEMOS IN PARALLEL")
    print("="*80)
    print("\nExecuting 4 code generation tasks simultaneously...\n")
    sys.stdout.flush()
    
    # Define all demo tasks
    def demo_rest_api():
        spec = {
            'entity': 'Product',
            'fields': [
                {'name': 'id', 'type': 'int', 'primary_key': True},
                {'name': 'name', 'type': 'str', 'required': True},
                {'name': 'description', 'type': 'str'},
                {'name': 'price', 'type': 'float', 'required': True},
                {'name': 'in_stock', 'type': 'bool', 'default': True},
                {'name': 'created_at', 'type': 'datetime', 'auto_now_add': True}
            ],
            'auth': True
        }
        return ('REST API', generator.generate_rest_api(spec, framework='fastapi'))
    
    def demo_react():
        spec = {
            'name': 'ProductList',
            'props': [
                {'name': 'categoryId', 'type': 'string', 'required': True},
                {'name': 'onProductClick', 'type': '(productId: string) => void'}
            ],
            'state': [
                {'name': 'products', 'type': 'Product[]'},
                {'name': 'loading', 'type': 'boolean'},
                {'name': 'error', 'type': 'Error | null'}
            ],
            'api_endpoint': '/api/products'
        }
        return ('React Component', generator.generate_react_component(spec))
    
    def demo_apex():
        spec = {
            'name': 'OpportunityService',
            'sobject': 'Opportunity',
            'operations': ['create', 'read', 'update', 'delete'],
            'trigger': True
        }
        return ('Apex Class', generator.generate_apex_class(spec))
    
    def demo_refactor():
        legacy_code = '''
def processUserData(userData):
    results = []
    for user in userData:
        if user[1] != None and user[2] != None:
            age = 2025 - int(user[2])
            if age >= 18:
                results.append(user[0] + " (" + str(age) + ")")
    return results
'''
        return ('Code Refactoring', generator.refactor_legacy_code(legacy_code, 'python', '3.10'))
    
    # Store results
    results = {}
    
    # Execute all demos in parallel
    start_time = time.time()
    
    with ThreadPoolExecutor(max_workers=4) as executor:
        # Submit all tasks
        futures = {
            executor.submit(demo_rest_api): '🔌 REST API',
            executor.submit(demo_react): '⚛️  React Component',
            executor.submit(demo_apex): '☁️  Apex Class',
            executor.submit(demo_refactor): '🔄 Refactoring'
        }
        
        # Process results as they complete
        for future in as_completed(futures):
            task_name = futures[future]
            try:
                name, result = future.result()
                results[name] = result
                print(f"✓ {task_name} completed")
                sys.stdout.flush()
            except Exception as e:
                print(f"✗ {task_name} failed: {e}")
                sys.stdout.flush()
    
    total_time = time.time() - start_time
    
    print(f"\n{'='*80}")
    print(f"✅ ALL DEMOS COMPLETED IN {total_time:.1f} SECONDS")
    print(f"{'='*80}\n")
    sys.stdout.flush()
    
    # Display summary
    print("📊 GENERATION SUMMARY:\n")
    
    for name, result in results.items():
        print(f"{name}:")
        if hasattr(result, 'metadata'):
            print(f"  Files: {len(result.files)}")
            print(f"  Lines: {sum(len(c.split('\n')) for c in result.files.values())}")
            print(f"  Cost: ${result.metadata['cost_usd']:.4f}")
            print(f"  Time: {result.metadata['latency_ms']:.0f}ms")
        else:
            # Refactoring result
            print(f"  Lines: {len(result.refactored_code.split('\n'))}")
            print(f"  Cost: ${result.metadata['cost_usd']:.4f}")
            print(f"  Improvements: {len(result.improvements)}")
        print()
    
    # Calculate total savings
    sequential_time = sum(r.metadata['latency_ms'] for r in results.values()) / 1000
    time_saved = sequential_time - total_time
    speedup = sequential_time / total_time
    
    print(f"⚡ PERFORMANCE:")
    print(f"  Sequential time: {sequential_time:.1f}s")
    print(f"  Parallel time: {total_time:.1f}s")
    print(f"  Time saved: {time_saved:.1f}s")
    print(f"  Speedup: {speedup:.1f}x faster")
    print()
    
    total_cost = sum(r.metadata['cost_usd'] for r in results.values())
    total_lines = sum(
        sum(len(c.split('\n')) for c in r.files.values()) if hasattr(r, 'files')
        else len(r.refactored_code.split('\n'))
        for r in results.values()
    )
    
    print(f"💰 TOTAL COST: ${total_cost:.4f}")
    print(f"📝 TOTAL LINES: {total_lines:,}")
    print(f"💵 COST PER 100 LINES: ${(total_cost / total_lines * 100):.4f}")
    
else:
    print("⚠️  Generator not available - configure API keys")
````
